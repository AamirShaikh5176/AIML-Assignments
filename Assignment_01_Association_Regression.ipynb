```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression

# Create sample dataset
data = {
    'Hours_Studied': [2, 3, 4, 5, 6, 7, 8],
    'Exam_Score': [45, 50, 55, 65, 70, 78, 85]
}

df = pd.DataFrame(data)

# Display dataset
print("Dataset:")
print(df)

# Identify independent and dependent variables
X = df[['Hours_Studied']]
Y = df['Exam_Score']

# Calculate correlation
correlation, p_value = pearsonr(df['Hours_Studied'], df['Exam_Score'])

print("\nCorrelation Coefficient:", correlation)
print("P-value:", p_value)

# Linear Regression
model = LinearRegression()
model.fit(X, Y)

# Predictions
df['Predicted_Score'] = model.predict(X)

print("\nRegression Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)

# Display predictions
print("\nPredicted Values:")
print(df)

# Scatter plot with regression line
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Hours_Studied', y='Exam_Score', data=df)
plt.plot(df['Hours_Studied'], df['Predicted_Score'])
plt.xlabel("Hours Studied (Independent Variable)")
plt.ylabel("Exam Score (Dependent Variable)")
plt.title("Association Between Hours Studied and Exam Score")
plt.show()
```
